```python
#> imports
import sys
import numpy as np
import matplotlib.pyplot as plt

#> import opus path
sys.path.insert(1, '../')
import lensing
import plot
```

```python
#> image positions
images = [[-8.379,  -15.899],
            [2.355,  -17.445],
            [-19.378,  -1.900],
            [6.343,  7.713],
            [0.500,  0.679]]
images = np.array(images)
```

```python
#> getting arival order
origin = (0.0, 0.0)
images_ordered = lensing.arrivalOrder(images, *origin)
print(images_ordered)
```

```python
#> initializing plot
fig, ax = plt.subplots(1,1,figsize=(6,6))
ax.grid(ls=':', alpha=0.5)
ax.set_xlabel(r'$\mathbf{\Delta RA ~~[arcsec]}$', fontsize=15, labelpad=10)
ax.set_ylabel(r'$\mathbf{\Delta DEC ~~[arcsec]}$', fontsize=15)
ax.invert_xaxis() 

#> plotting!
plot.plotArrival(ax, images_ordered)
ax.scatter(*origin, marker='+', c='k', s=100)

plt.show()
```
