```python
#> imports
import numpy as np
import matplotlib.pyplot as plt
import scipy.stats as stats
```

### Galaxy Model Parameters
To generate a galaxy model from the simulator, a galaxy needs values for the following parameters:

$
\begin{align}
M_{NFW} &= \log_{10}(M/M_{\odot}) = \textup{mass of NFW} \\
c &= \textup{concentration of NFW} \\
q_{NFW} &= \textup{axis-ratio of NFW} \\
\theta_{NFW} &= \textup{position angle of NFW} \\ \\
M_{Hern} &= \log_{10}(M/M_{\odot}) = \textup{mass of Hernquist} \\
R_E &= \textup{effective radius of Hernquist} \\
q_{Hern} &= \textup{axis-ratio of Hernquist} \\ \\
a_1 &= \textup{normalization of m=1} \\
a_3 &= \textup{normalization of m=3} \\
a_4 &= \textup{normalization of m=4} \\
\theta_1 &= \textup{position angle of m=1} \\
\theta_3 &= \textup{position angle of m=3} \\
\theta_4 &= \textup{position angle of m=4} \\ \\
\gamma_{EX} &= \textup{normalization of external shear} \\
\theta_{\gamma} &= \textup{position angle of external shear} \\ \\
z_l &= \textup{lens redshift} \\
z_s &= \textup{source redshift}
\end{align}
$

### Source Parameters
To lens a source and obtain a quad, a source needs values for the following parameters:

$
\begin{align}
x_s &= \textup{x-position of the source} \\
y_s &= \textup{y-position of the source}
\end{align}
$

## **Axis-ratio ($q_\bullet, q_*$)**

```python
lower_bound = 0.4
upper_bound = 1
scale = upper_bound - lower_bound
x = np.linspace(lower_bound, upper_bound, 100)

a = 10
b = 2
beta = stats.beta.pdf(x, a, b, loc=lower_bound, scale=scale)

fig, ax = plt.subplots(1,1,figsize=(6,6))
ax.grid(ls=':', alpha=0.3)
ax.set_xlabel('axis-ratio; q [-]', fontweight='bold', labelpad=10)
ax.plot(x, beta)
plt.show()
```

## **Position Angle ($\theta_\bullet$)**

```python
lower_bound = -90
upper_bound = 90
scale = upper_bound - lower_bound
x = np.linspace(lower_bound, upper_bound, 100)

a = 30
b = 30
beta = stats.beta.pdf(x, a, b, loc=lower_bound, scale=scale)
normal = stats.norm.pdf(x, loc=0, scale=10)

fig, ax = plt.subplots(1,1,figsize=(6,6))
ax.grid(ls=':', alpha=0.3)
ax.set_xlabel('axis-ratio; q [-]', fontweight='bold', labelpad=10)
ax.plot(x, beta)
# ax.plot(x, normal)
plt.show()
```
