```python
#> imports
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
```

#### ***Loading Data***
---

```python
#> file declarations
m1_file = '../data/baseline/2607021145q75_m1_t1_fit.npy'
offset_file = '../data/baseline/2607021204q75_offset_300pc.npy'
offset_file = '../data/baseline/2607081305q75_offset_600pc_uniform_1000g_100s.npy'

#> loading data
m1_data = np.load(m1_file, allow_pickle=True).item()
offset_data = np.load(offset_file, allow_pickle=True).item()

#> printing keys
print('M1 Keys:', m1_data.keys())
print('Offset Keys:', offset_data.keys())

#> declarations
observables = ['t12', 't23', 't34', 'd2/d1', 'd3/d1' ,'d4/d1', 'dt23']
```

#### ***Parsing Offset/ M1***
---

```python
#> converts cartesian to polar
def to_polar(x,y):
    return(np.sqrt(x**2 + y**2), np.rad2deg(np.arctan2(y, x)))

numSource_gal = 100

#> parsing offset
offset = []
for i, profile in enumerate(offset_data['bprofiles']):
    for j in range(numSource_gal):
        offset.append(to_polar(profile['nfw'][0]['x0'], profile['nfw'][0]['y0']))
offset = np.array(offset)

#> parsing m1
m1 = []
for i, profile in enumerate(m1_data['bprofiles']):
    m1.append([profile['mult'][0]['norm'], profile['mult'][0]['theta']])
m1 = np.array(m1)

#> combining into dataframe
off_df = pd.DataFrame(offset_data['im_obs'], columns=observables)
off_df[['offset_r', 'offset_t']] = offset # (arcsec, degree)
off_df

#> combining into dataframe
m1_df = pd.DataFrame(m1_data['im_obs'], columns=observables)
m1_df[['m1_n', 'm1_t']] = m1 # (arcsec, degree)
m1_df
```

```python
#> declarations
fontsize = 15
labelpad = 10

#> initializing plot
fig, ax = plt.subplots(1,1,figsize=(6,6))
ax.grid(ls=':', alpha=0.5)
ax.set_xlabel('Radial Offset [arcsec]', fontsize=fontsize, fontweight='bold', labelpad=labelpad)
ax.set_ylabel(r'$\mathbf{\Delta\theta_{23}}$', fontsize=fontsize, labelpad=labelpad)
ax.set_title(r'$\mathbf{\Delta\theta_{23}} ~vs. Offset$', fontsize=fontsize, fontweight='bold')

#> plotting!
ax.scatter(off_df['offset_r'], off_df['dt23'], c='k', marker='.', alpha=0.05)

plt.show(); plt.close()

### NEED TO HISTOGRAM THESE, THEN PLOT RMS DELTA THETA 23 VERSUS OFFSET
```
