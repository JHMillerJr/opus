#### ***Intro to Newton's Method***
---

The following is taken/ summarized from [Wikipedia](https://en.wikipedia.org/wiki/Newton%27s_method).

Newton's Method (or the Newton-Raphson method) is a "root-finding algorithm which produces successively better approximations to the roots (or zeros) of a real-valued function." Take a real-valued function $f$, its derivative $f^\prime$, and an initial guess $x_0$ for a roof of $f$. If the inital guess $x_0$ is sufficicently close, then

$$
x_1 = x_0 - \frac{f(x_0)}{f^\prime(x_0)}
$$

is a better approximation of the root than $x_0$. That is, $x_1$ is the x-intercept of the line starting at $x_0$ with slope $f^\prime(x_0)$. The processes can be iterated upon until a sufficient approximation of the root is acquired: 

$$
x_{n+1} = x_n - \frac{f(x_n)}{f^\prime(x_n)}
$$

### ***1D Example***
---
Take the equation $f(x) = x^2 - 2$. Say we want to find the zero of $f(x)$ with Newton's method. Let's start with an initial guess $x_0 = 1$ and iterate for $N=5$.

```python
#> functions
f = lambda x: x**2 - 2
f_prime = lambda x: 2*x

#> declarations
N = 5
x0 = 1

#> initial eval
f_n, x = [f(x0)], [x0]

#> iterating
for n in range(N):

    #> newton's method
    x_n = x[n]
    x_n_1 = x_n - ( f(x_n) / f_prime(x_n) )

    #> appending
    f_n.append(f(x_n_1))
    x.append(x_n_1)

    #> printing
    print(f'> Step:{n+1}, f(x_(n+1))={f(x_n_1):.2e}, and x_(n+1)={x_n_1}')
```
